```python
print("Shape before removing missing values:", df.shape)

df_drop = df.dropna()

print("Shape after removing missing values:", df_drop.shape)
```

```python
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

df_mean["Age"].isnull().sum()  # verifying no missing values remain
```

```python
df_median = df.copy()
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

df_median["Age"].isnull().sum()  # verifying no missing values remain
```

```python
df_mode = df.copy()
df_mode["Embarked"] = df_mode["Embarked"].fillna(df_mode["Embarked"].mode()[0])

df_mode["Embarked"].isnull().sum()  # verifying no missing values remain
```
