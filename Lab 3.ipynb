```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Dataset
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct"],
    "Copies_Sold": [420, 450, 480, 510, 530, 570, 600, 650, 670, 690]
}

df = pd.DataFrame(data)

df
```

```python
# Prepare data for the model
X = np.arange(1, 11).reshape(-1, 1)
y = df["Copies_Sold"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

print("Model trained successfully!")
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Predict sales for the existing months
y_pred = model.predict(X)

print("Predicted Sales:")
print(y_pred)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, y_pred)

print("R² Score:", r2)
```

```python
# Predict sales for November and December

future_months = np.array([[11], [12]])

future_predictions = model.predict(future_months)

print("November prediction:", future_predictions[0])
print("December prediction:", future_predictions[1])
```

```python
# Visualize actual and predicted sales

plt.figure(figsize=(10, 5))

plt.scatter(X, y, label="Actual Sales")
plt.plot(X, y_pred, label="Regression Line")

plt.scatter(
    future_months,
    future_predictions,
    label="Future Predictions"
)

plt.xlabel("Month")
plt.ylabel("Copies Sold")
plt.title("Monthly Newspaper Sales Prediction")
plt.legend()
plt.grid(True)

plt.show()
```

## Conclusion

The Monthly Newspaper Sales dataset was analyzed using Linear Regression. The model showed a strong relationship between month and copies sold. Future sales were predicted for November and December, showing an expected increase in newspaper sales.
