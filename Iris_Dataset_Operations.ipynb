```python
print("Shape before removing missing values:", df.shape)

df_drop = df.dropna()

print("Shape after removing missing values:", df_drop.shape)
```

```python
df_demo = df.copy()

# Manually introducing a few missing values for demonstration purposes only
df_demo.loc[0:2, "sepal_length"] = None
df_demo.loc[5:6, "species"] = None

df_demo.isnull().sum()
```

```python
df_mean = df_demo.copy()
df_mean["sepal_length"] = df_mean["sepal_length"].fillna(df_mean["sepal_length"].mean())

df_mean["sepal_length"].isnull().sum()  # verifying no missing values remain
```

```python
df_median = df_demo.copy()
df_median["sepal_length"] = df_median["sepal_length"].fillna(df_median["sepal_length"].median())

df_median["sepal_length"].isnull().sum()  # verifying no missing values remain
```

```python
df_mode = df_demo.copy()
df_mode["species"] = df_mode["species"].fillna(df_mode["species"].mode()[0])

df_mode["species"].isnull().sum()  # verifying no missing values remain
```
